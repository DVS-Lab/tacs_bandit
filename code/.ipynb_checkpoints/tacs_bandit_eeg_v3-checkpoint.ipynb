# tACS Bandit EEG Analyses (v3)

Tier 2 EEG analysis notebook for: *Examining the effects of theta-tACS over left DLPFC on reward-based learning across the adult lifespan*

**v3 Updates (based on diagnostic review):**
- Fixed artifact rejection: now applies threshold to high-pass filtered data, not raw amplitude
- Reference-aware processing: separate pipelines for earclip vs. no-earclip participants
- Adaptive stimulation detection: threshold scaled relative to participant's baseline 6 Hz power
- Added protocol compliance flags and notes for known issues
- Improved signal quality diagnostics

**Known Data Issues (from v2 review):**
- sub-10998: Missing Run 1; Run 6 administered as sham (should have been active)
- sub-11773: Runs 6-7 administered as sham (should have been active)
- sub-10951: Electrode artifact (rectangular jumps) — likely loose connection
- First 6 participants recorded without earclip reference

**Key Constraints:**
- Stimulation channels (F3, Fp1, FCz, FT7) blanked by NIC2 on all runs
- Only 3 EEG-only channels available (F4, P4, P3)
- LSL task markers not reaching NIC2 trigger channel

## 1. Data Loading Utilities

```python
# =============================================================================
# 1. NIC2 File Loading
# =============================================================================

def load_nic2_run(easy_path, info_path=None, imp_path=None):
    """
    Load a single NIC2 recording.
    
    .easy file structure (tab-delimited, no header):
        Columns 0-7:  8 EEG channels in nanovolts (nV)
        Column 11:    Trigger/marker channel
        Column 12:    Timestamp (Unix epoch, milliseconds)
    """
    raw = np.loadtxt(easy_path)
    eeg = raw[:, :8]
    triggers = raw[:, 11] if raw.shape[1] > 11 else np.zeros(len(raw))
    timestamps = raw[:, 12] if raw.shape[1] > 12 else np.arange(len(raw))
    
    # Parse .info for metadata
    info = {}
    if info_path and Path(info_path).exists():
        with open(info_path, 'r') as f:
            info_text = f.read()
        for line in info_text.split('\n'):
            if 'EEG sampling rate' in line and 'Samples' in line:
                info['fs'] = int(line.split(':')[1].strip().split()[0])
            if 'Step name' in line:
                info['step_name'] = line.split(':')[1].strip()
            if 'Atacs (uA)' in line:
                val = int(line.split(':')[1].strip())
                if val > 0:
                    info['has_active_stim'] = True
    
    fs = info.get('fs', FS)
    stim_mask = (eeg[:, 0] == -1)
    
    # Load impedance from _IMP.txt if available
    impedance = None
    if imp_path and Path(imp_path).exists():
        try:
            imp_raw = np.loadtxt(imp_path)
            if imp_raw.ndim == 2 and imp_raw.shape[0] > 0:
                impedance = {
                    'values': imp_raw[:, :8],
                    'timestamps': imp_raw[:, 8] if imp_raw.shape[1] > 8 else None,
                    'mean_per_channel': np.mean(imp_raw[:, :8], axis=0),
                }
        except Exception:
            pass
    
    return {
        'eeg': eeg,
        'timestamps': timestamps,
        'triggers': triggers,
        'fs': fs,
        'stim_mask': stim_mask,
        'impedance': impedance,
        'info': info,
    }


def discover_runs(eeg_dir, subject_id):
    """
    Find all NIC2 files for a subject and organize by run number.
    Handles duplicate files by preferring the longest recording.
    """
    pattern = str(eeg_dir / f'*sub-{subject_id}_Run*')
    all_files = glob.glob(pattern + '.easy')
    
    runs = {}
    for easy_path in sorted(all_files):
        easy_path = Path(easy_path)
        fname = easy_path.stem
        
        try:
            run_str = fname.split('Run')[1].strip().split('_')[0].strip()
            run_num = int(run_str)
        except (IndexError, ValueError):
            continue
        
        info_path = easy_path.with_suffix('.info')
        imp_candidates = list(eeg_dir.glob(f'*sub-{subject_id}_Run*{run_num}_IMP.txt'))
        imp_path = imp_candidates[0] if imp_candidates else None
        
        file_size = easy_path.stat().st_size
        
        if run_num not in runs or file_size > runs[run_num]['size']:
            runs[run_num] = {
                'easy': easy_path,
                'info': info_path if info_path.exists() else None,
                'imp': imp_path,
                'size': file_size,
            }
    
    for run in runs.values():
        del run['size']
    
    return runs


print('Loading utilities ready.')
```

## 2. Preprocessing Functions (Reference-Aware)

Key changes from v2:
- Artifact rejection now operates on **high-pass filtered** data, not raw amplitude
- Separate parameter sets for earclip vs. no-earclip participants
- Added electrode pop detection (catches sudden jumps like sub-10951)

```python
# =============================================================================
# 2. Preprocessing Functions
# =============================================================================

def apply_filters(eeg_data, fs=FS, highpass=0.5, lowpass=40.0, notch=60.0):
    """
    Apply bandpass and notch filters to EEG data.
    
    Parameters
    ----------
    eeg_data : ndarray (samples,) or (samples, channels)
        Raw EEG in nanovolts
    fs : float
        Sampling rate
    highpass : float
        High-pass cutoff (Hz). Removes DC offset and slow drift.
    lowpass : float
        Low-pass cutoff (Hz). Removes high-frequency noise.
    notch : float or None
        Notch filter frequency (Hz) for line noise. None to skip.
    
    Returns
    -------
    filtered : ndarray, same shape as input
    """
    nyq = fs / 2
    
    # Bandpass filter
    b, a = butter(4, [highpass / nyq, lowpass / nyq], btype='band')
    
    # Handle 1D vs 2D input
    if eeg_data.ndim == 1:
        filtered = filtfilt(b, a, eeg_data.astype(float))
    else:
        filtered = np.zeros_like(eeg_data, dtype=float)
        for ch in range(eeg_data.shape[1]):
            filtered[:, ch] = filtfilt(b, a, eeg_data[:, ch].astype(float))
    
    # Notch filter for line noise
    if notch is not None:
        b_notch, a_notch = iirnotch(notch, Q=30, fs=fs)
        if filtered.ndim == 1:
            filtered = filtfilt(b_notch, a_notch, filtered)
        else:
            for ch in range(filtered.shape[1]):
                filtered[:, ch] = filtfilt(b_notch, a_notch, filtered[:, ch])
    
    return filtered


def avg_rereference(eeg, ch_indices=EEG_CH_IDX):
    """
    Average re-reference across specified channels.
    Required for no-earclip participants to remove common-mode signal.
    """
    subset = eeg[:, ch_indices].astype(float)
    avg = np.mean(subset, axis=1, keepdims=True)
    return subset - avg


def reject_artifacts(eeg_data, fs=FS, threshold_uv=150, 
                     highpass=0.5, detect_pops=True, pop_threshold_uv=500):
    """
    Artifact rejection on FILTERED data.
    
    v3 FIX: Previous version applied threshold to raw data with DC offsets,
    causing 95%+ rejection rates. Now we high-pass filter first to remove
    the DC offset, then apply the threshold.
    
    Parameters
    ----------
    eeg_data : ndarray (samples,)
        EEG data in nanovolts (will be converted to µV internally)
    fs : float
        Sampling rate
    threshold_uv : float
        Amplitude threshold in µV (applied to filtered data)
    highpass : float
        High-pass cutoff for DC removal before thresholding
    detect_pops : bool
        If True, also flag electrode pops (sudden jumps)
    pop_threshold_uv : float
        Threshold for sample-to-sample jumps (µV)
    
    Returns
    -------
    clean_mask : bool array
        True for clean samples
    pct_rejected : float
        Percentage of samples rejected
    rejection_info : dict
        Details about what was rejected
    """
    # Convert to µV
    eeg_uv = eeg_data / 1000
    
    # High-pass filter to remove DC offset before thresholding
    nyq = fs / 2
    b, a = butter(2, highpass / nyq, btype='high')
    eeg_filtered = filtfilt(b, a, eeg_uv.astype(float))
    
    # Amplitude threshold on filtered data
    amp_clean = np.abs(eeg_filtered) < threshold_uv
    
    # Electrode pop detection (sudden jumps)
    if detect_pops:
        diff = np.abs(np.diff(eeg_uv))
        pop_mask = np.concatenate([[False], diff > pop_threshold_uv])
        # Extend pop rejection by 50ms on each side
        extend_samples = int(0.05 * fs)
        pop_indices = np.where(pop_mask)[0]
        for idx in pop_indices:
            start = max(0, idx - extend_samples)
            end = min(len(pop_mask), idx + extend_samples)
            pop_mask[start:end] = True
        pop_clean = ~pop_mask
    else:
        pop_clean = np.ones(len(eeg_uv), dtype=bool)
    
    # Combine masks
    clean_mask = amp_clean & pop_clean
    pct_rejected = 100 * (1 - clean_mask.mean())
    
    rejection_info = {
        'pct_amplitude': 100 * (1 - amp_clean.mean()),
        'pct_pops': 100 * (1 - pop_clean.mean()) if detect_pops else 0,
        'pct_total': pct_rejected,
    }
    
    return clean_mask, pct_rejected, rejection_info


def preprocess_eeg(eeg_raw, has_earclip, fs=FS, return_intermediate=False):
    """
    Full preprocessing pipeline with reference-aware settings.
    
    Parameters
    ----------
    eeg_raw : ndarray (samples, 8)
        Raw EEG from NIC2 (all 8 channels, nanovolts)
    has_earclip : bool
        Whether this participant had earclip reference
    fs : float
        Sampling rate
    return_intermediate : bool
        If True, return intermediate processing stages for debugging
    
    Returns
    -------
    eeg_clean : ndarray (samples, 3)
        Preprocessed EEG for the 3 recording channels (F4, P4, P3)
    clean_mask : bool array
        Which samples are clean
    preproc_info : dict
        Preprocessing details
    """
    params = PREPROC_PARAMS['earclip' if has_earclip else 'no_earclip']
    
    # Step 1: Extract EEG-only channels
    eeg_subset = eeg_raw[:, EEG_CH_IDX].astype(float)
    
    # Step 2: Average re-reference (only for no-earclip)
    if params['apply_avg_reref']:
        eeg_reref = avg_rereference(eeg_raw)
    else:
        eeg_reref = eeg_subset
    
    # Step 3: Bandpass + notch filter
    eeg_filtered = apply_filters(
        eeg_reref, fs=fs,
        highpass=params['highpass_cutoff'],
        lowpass=params['lowpass_cutoff'],
        notch=params['notch_freq']
    )
    
    # Step 4: Artifact rejection (on F4, primary channel)
    clean_mask, pct_rejected, rejection_info = reject_artifacts(
        eeg_filtered[:, 0],  # F4
        fs=fs,
        threshold_uv=params['artifact_threshold_uv'],
        highpass=params['highpass_cutoff'],
    )
    
    preproc_info = {
        'has_earclip': has_earclip,
        'params': params,
        'rejection': rejection_info,
        'n_samples_clean': np.sum(clean_mask),
        'n_samples_total': len(clean_mask),
    }
    
    if return_intermediate:
        preproc_info['eeg_subset'] = eeg_subset
        preproc_info['eeg_reref'] = eeg_reref
    
    return eeg_filtered, clean_mask, preproc_info


print('Preprocessing functions ready.')
```

## 3. Stimulation Detection (Adaptive Threshold)

v3 FIX: Previous version used a fixed 120 dB threshold, which worked for no-earclip participants but missed active stimulation in earclip participants (who have lower overall power). Now we use a **relative threshold**: active stimulation is detected if 6 Hz power in the middle of the run exceeds the participant's baseline by X dB.

```python
# =============================================================================
# 3. Stimulation Detection
# =============================================================================

def compute_6hz_timecourse(eeg_data, fs=FS):
    """
    Compute time-resolved 6 Hz power across a run.
    
    Returns
    -------
    t_spec : array
        Time points (seconds)
    power_db : array
        6 Hz power in dB at each time point
    """
    f, t_spec, Sxx = signal.spectrogram(
        signal.detrend(eeg_data), fs=fs,
        nperseg=1000, noverlap=750, scaling='density'
    )
    
    six_hz_mask = (f >= 5.5) & (f <= 6.5)
    power_db = 10 * np.log10(np.mean(Sxx[six_hz_mask, :], axis=0) + 1e-30)
    
    return t_spec, power_db


def detect_active_stim_adaptive(eeg_data, fs=FS, threshold_db=15):
    """
    Detect active tACS using adaptive (relative) threshold.
    
    Instead of a fixed dB threshold, we compare the middle of the run
    to the edges (which should be near baseline even for active runs
    due to ramp-up/down).
    
    Parameters
    ----------
    eeg_data : ndarray
        EEG data for one channel (nanovolts)
    fs : float
        Sampling rate
    threshold_db : float
        How much higher the middle must be than edges to count as active
    
    Returns
    -------
    is_active : bool
        True if sustained active stimulation detected
    detection_info : dict
        Details about the detection
    """
    t_spec, power_db = compute_6hz_timecourse(eeg_data, fs)
    
    if len(t_spec) == 0:
        return False, {'error': 'No spectrogram computed'}
    
    duration = t_spec[-1]
    
    # Define time windows
    # Early: first 45 seconds (includes sham ramp)
    # Middle: 60-300 seconds (core of the run)
    # Late: last 45 seconds
    early_mask = t_spec <= 45
    middle_mask = (t_spec >= 60) & (t_spec <= min(300, duration - 45))
    late_mask = t_spec >= (duration - 45)
    
    # Compute mean power in each window
    early_power = np.mean(power_db[early_mask]) if np.any(early_mask) else np.nan
    middle_power = np.mean(power_db[middle_mask]) if np.any(middle_mask) else np.nan
    late_power = np.mean(power_db[late_mask]) if np.any(late_mask) else np.nan
    
    # Baseline estimate: use late window (after ramp-down for sham)
    # For active runs, late is still elevated, but middle should be similar
    # For sham runs, late should be back to baseline
    
    # Detection logic:
    # Active: middle is elevated AND stays elevated through late
    # Sham: early might be elevated (ramp), but middle and late are near baseline
    
    # Use early power as baseline reference for sham detection
    # But for active detection, compare middle to a true baseline estimate
    
    # Simple heuristic: if middle > 100 dB absolute AND middle is within 5 dB of early
    # (meaning it's sustained, not a brief ramp), call it active
    
    # More robust: compare middle to late
    # Active: middle ≈ late (both elevated)
    # Sham: middle < late by a lot (middle is baseline, late might have some ramp artifact)
    # Actually for sham: early is elevated, middle is LOW, late is low
    # For active: early is ramping up, middle is HIGH, late is ramping down but still high
    
    # Best approach: middle power significantly above a LOW baseline
    # Estimate baseline from the minimum of the three windows
    baseline_estimate = min(early_power, middle_power, late_power)
    
    # For sham: baseline_estimate ≈ middle_power ≈ late_power
    # For active: baseline_estimate << middle_power
    
    elevation = middle_power - baseline_estimate
    is_active = elevation > threshold_db and middle_power > 90  # Must also exceed absolute floor
    
    detection_info = {
        'early_power_db': early_power,
        'middle_power_db': middle_power,
        'late_power_db': late_power,
        'baseline_estimate_db': baseline_estimate,
        'elevation_db': elevation,
        'threshold_db': threshold_db,
        'duration_s': duration,
    }
    
    return is_active, detection_info


def classify_run_stimulation(eeg_data, fs=FS, has_earclip=True):
    """
    Classify a run as ACTIVE, SHAM, or NONE based on 6 Hz timecourse.
    
    Returns
    -------
    classification : str
        'ACTIVE', 'SHAM', or 'NONE'
    info : dict
        Detection details
    """
    params = PREPROC_PARAMS['earclip' if has_earclip else 'no_earclip']
    threshold = params['stim_detect_threshold_db']
    
    is_active, info = detect_active_stim_adaptive(eeg_data, fs, threshold)
    
    if is_active:
        classification = 'ACTIVE'
    elif info.get('early_power_db', 0) > info.get('middle_power_db', 0) + 10:
        # Early power elevated but middle is not → sham ramp
        classification = 'SHAM'
    else:
        classification = 'NONE'
    
    info['classification'] = classification
    return classification, info


print('Stimulation detection ready.')
```

## 4. Spectral Parameterization (ITF Extraction)

```python
# =============================================================================
# 4. Spectral Parameterization
# =============================================================================

def compute_psd(eeg_data, fs=FS, nperseg=2048):
    """
    Compute power spectral density using Welch's method.
    
    Expects ALREADY PREPROCESSED data (filtered, artifact-rejected).
    """
    # Ensure we have enough samples
    if len(eeg_data) < nperseg:
        nperseg = len(eeg_data)
        if nperseg < 256:
            return None, None  # Not enough data
    
    freqs, psd = signal.welch(
        signal.detrend(eeg_data.astype(float)),
        fs=fs, nperseg=nperseg, noverlap=nperseg//2
    )
    return freqs, psd


def fit_specparam(freqs, psd_log, freq_range=(1, 40)):
    """
    Fit a specparam model to a log-power PSD.
    
    Parameters
    ----------
    freqs : array
        Frequency axis
    psd_log : array
        Log10 power spectral density
    freq_range : tuple
        Frequency range to fit
    
    Returns
    -------
    model : SpectralModel
        Fitted model
    """
    sm = SpectralModel(
        peak_width_limits=(0.5, 4),   # Theta peaks are ~1-3 Hz wide
        max_n_peaks=4,
        min_peak_height=0.15,
        aperiodic_mode='fixed',
        verbose=False,
    )
    
    # Check for invalid values
    if np.any(np.isnan(psd_log)) or np.any(np.isinf(psd_log)):
        return None
    
    try:
        sm.fit(freqs, psd_log, freq_range)
        return sm
    except Exception:
        return None


def extract_itf(model, theta_range=(4, 8)):
    """
    Extract individual theta frequency from fitted specparam model.
    """
    if model is None:
        return {'itf': None, 'power': None, 'bandwidth': None, 'has_theta': False}
    
    try:
        peaks = model.results.params.periodic.params
    except Exception:
        return {'itf': None, 'power': None, 'bandwidth': None, 'has_theta': False}
    
    if peaks.ndim == 1:
        if len(peaks) == 0:
            return {'itf': None, 'power': None, 'bandwidth': None, 'has_theta': False}
        peaks = peaks.reshape(1, -1)
    
    if len(peaks) == 0:
        return {'itf': None, 'power': None, 'bandwidth': None, 'has_theta': False}
    
    # Find peaks in theta range
    theta_peaks = peaks[(peaks[:, 0] >= theta_range[0]) &
                        (peaks[:, 0] <= theta_range[1])]
    
    if len(theta_peaks) == 0:
        return {'itf': None, 'power': None, 'bandwidth': None, 'has_theta': False}
    
    # Return the strongest theta peak
    best = theta_peaks[np.argmax(theta_peaks[:, 1])]
    return {
        'itf': best[0],
        'power': best[1],
        'bandwidth': best[2],
        'has_theta': True,
    }


def get_model_fit_quality(model):
    """Extract R² from fitted model."""
    if model is None:
        return None
    try:
        return model.results.metrics.results['gof_rsquared']
    except Exception:
        return None


print('Spectral parameterization ready.')
```

## 5. Single-Subject Processing Pipeline

```python
# =============================================================================
# 5. Single-Subject Processing
# =============================================================================

def process_subject(subject_id, eeg_dir=EEG_DIR, verbose=True):
    """
    Full processing pipeline for one subject.
    
    Returns
    -------
    results : dict
        Subject-level results including:
        - Protocol compliance for each run
        - ITF estimates from baseline runs
        - Signal quality metrics
    """
    sub_info = SUBJECT_INFO[subject_id]
    cb_order = sub_info['counterbalance']
    has_earclip = sub_info['earclip']
    condition_map = RUN_CONDITIONS[cb_order]
    
    run_files = discover_runs(eeg_dir, subject_id)
    
    if len(run_files) == 0:
        return {'subject_id': subject_id, 'error': 'No runs found'}
    
    results = {
        'subject_id': subject_id,
        'counterbalance': cb_order,
        'has_earclip': has_earclip,
        'notes': sub_info.get('notes', ''),
        'n_runs': len(run_files),
        'runs': {},
        'protocol_mismatches': [],
        'itf_estimates': {'F4': [], 'P4': [], 'P3': []},
    }
    
    for run_num in sorted(run_files.keys()):
        files = run_files[run_num]
        expected_condition = condition_map.get(run_num, 'unknown')
        
        if verbose:
            print(f'  Run {run_num} ({expected_condition})...', end=' ')
        
        try:
            nic2 = load_nic2_run(files['easy'], files['info'], files['imp'])
        except Exception as e:
            results['runs'][run_num] = {'error': str(e)}
            if verbose:
                print(f'ERROR: {e}')
            continue
        
        # Preprocess
        eeg_filtered, clean_mask, preproc_info = preprocess_eeg(
            nic2['eeg'], has_earclip, nic2['fs']
        )
        
        # Detect stimulation
        stim_class, stim_info = classify_run_stimulation(
            nic2['eeg'][:, EEG_CH_IDX[0]],  # F4, raw for stim detection
            nic2['fs'],
            has_earclip
        )
        
        # Check protocol compliance
        expected_stim = 'ACTIVE' if expected_condition == 'active' else 'NONE'
        if expected_condition == 'sham':
            expected_stim = 'SHAM'  # or NONE, both acceptable
        
        # Mismatch if expected active but got something else
        protocol_ok = True
        if expected_condition == 'active' and stim_class != 'ACTIVE':
            protocol_ok = False
            results['protocol_mismatches'].append(run_num)
        
        run_result = {
            'condition': expected_condition,
            'stim_detected': stim_class,
            'protocol_ok': protocol_ok,
            'duration_s': stim_info.get('duration_s', 0),
            'pct_rejected': preproc_info['rejection']['pct_total'],
            'n_clean_samples': preproc_info['n_samples_clean'],
            'stim_info': stim_info,
        }
        
        # Extract ITF from baseline runs
        if expected_condition == 'baseline' and stim_class != 'ACTIVE':
            # Get clean data
            eeg_clean = eeg_filtered[clean_mask]
            
            if len(eeg_clean) >= 512:  # Minimum samples for PSD
                for ch_idx, ch_label in enumerate(EEG_CH_LABELS):
                    freqs, psd = compute_psd(eeg_clean[:, ch_idx])
                    
                    if freqs is not None:
                        psd_log = 10 * np.log10(psd + 1e-30)
                        model = fit_specparam(freqs, psd_log)
                        itf_result = extract_itf(model)
                        r_squared = get_model_fit_quality(model)
                        
                        run_result[f'itf_{ch_label}'] = itf_result['itf']
                        run_result[f'itf_{ch_label}_power'] = itf_result['power']
                        run_result[f'itf_{ch_label}_r2'] = r_squared
                        
                        # Store for aggregation if good fit
                        if itf_result['has_theta'] and r_squared is not None and r_squared >= 0.85:
                            results['itf_estimates'][ch_label].append(itf_result['itf'])
        
        results['runs'][run_num] = run_result
        
        if verbose:
            status = '✓' if protocol_ok else '⚠'
            rej_str = f'{run_result["pct_rejected"]:.1f}% rej'
            print(f'{status} {stim_class}, {rej_str}')
    
    # Aggregate ITF across baseline runs
    for ch_label in EEG_CH_LABELS:
        estimates = results['itf_estimates'][ch_label]
        if len(estimates) > 0:
            results[f'itf_{ch_label}'] = np.median(estimates)
            results[f'itf_{ch_label}_n'] = len(estimates)
            results[f'itf_{ch_label}_range'] = np.ptp(estimates) if len(estimates) > 1 else 0
        else:
            results[f'itf_{ch_label}'] = None
            results[f'itf_{ch_label}_n'] = 0
    
    return results


print('Single-subject pipeline ready.')
```

## 6. Process Single Subject (Interactive)

```python
# =============================================================================
# 6. Interactive Single-Subject Analysis
# =============================================================================

SUBJECT_ID = '10636'  # Change this to process a different subject

print(f'Processing sub-{SUBJECT_ID}...')
print('=' * 60)

result = process_subject(SUBJECT_ID, verbose=True)

print()
print('=' * 60)
print(f'Summary for sub-{SUBJECT_ID}:')
print(f'  Counterbalance: {result["counterbalance"]}')
print(f'  Earclip: {result["has_earclip"]}')
print(f'  Runs processed: {result["n_runs"]}')
print(f'  Protocol mismatches: {result["protocol_mismatches"] if result["protocol_mismatches"] else "None"}')
print()
print('ITF Estimates (F4):')
if result.get('itf_F4') is not None:
    print(f'  ITF: {result["itf_F4"]:.2f} Hz (n={result["itf_F4_n"]})')
    print(f'  Distance from 6 Hz tACS: {abs(result["itf_F4"] - 6):.2f} Hz')
else:
    print('  Not detected')

if result.get('notes'):
    print(f'\n⚠ Note: {result["notes"]}')
```

## 7. Batch Process All Subjects

```python
# =============================================================================
# 7. Batch Processing
# =============================================================================

all_results = []

print('Processing all subjects...')
print('=' * 70)

for subject_id in SUBJECT_INFO.keys():
    print(f'\nsub-{subject_id}')
    result = process_subject(subject_id, verbose=True)
    all_results.append(result)

print('\n' + '=' * 70)
print('Batch processing complete.')
```

```python
# =============================================================================
# 7b. Group Summary Table
# =============================================================================

summary_rows = []

for result in all_results:
    if 'error' in result:
        summary_rows.append({
            'subject_id': result['subject_id'],
            'status': 'ERROR',
        })
        continue
    
    # Compute mean rejection rate across runs
    rej_rates = [r.get('pct_rejected', 0) for r in result['runs'].values() 
                 if 'pct_rejected' in r]
    mean_rej = np.mean(rej_rates) if rej_rates else np.nan
    
    summary_rows.append({
        'subject_id': result['subject_id'],
        'counterbalance': result['counterbalance'],
        'earclip': result['has_earclip'],
        'n_runs': result['n_runs'],
        'protocol_ok': len(result['protocol_mismatches']) == 0,
        'mismatches': str(result['protocol_mismatches']) if result['protocol_mismatches'] else '',
        'mean_pct_rejected': mean_rej,
        'itf_F4': result.get('itf_F4'),
        'itf_F4_n': result.get('itf_F4_n', 0),
        'notes': result.get('notes', ''),
    })

summary_df = pd.DataFrame(summary_rows)

print('\nGroup Summary')
print('=' * 100)
print(summary_df.to_string(index=False))
print()

# ITF statistics
valid_itf = summary_df.dropna(subset=['itf_F4'])

if len(valid_itf) > 0:
    print('\nITF Statistics (F4, baseline runs):')
    print(f'  N with valid ITF: {len(valid_itf)} / {len(summary_df)}')
    print(f'  Mean ITF: {valid_itf["itf_F4"].mean():.2f} Hz')
    print(f'  SD: {valid_itf["itf_F4"].std():.2f} Hz')
    print(f'  Range: {valid_itf["itf_F4"].min():.2f} – {valid_itf["itf_F4"].max():.2f} Hz')
    
    # By earclip status
    for earclip_status in [True, False]:
        subset = valid_itf[valid_itf['earclip'] == earclip_status]['itf_F4']
        label = 'Earclip' if earclip_status else 'No earclip'
        if len(subset) > 0:
            print(f'  {label} (n={len(subset)}): {subset.mean():.2f} ± {subset.std():.2f} Hz')
else:
    print('\nNo valid ITF estimates found.')

# Artifact rejection summary
print('\nArtifact Rejection Rates:')
print(f'  Mean across subjects: {summary_df["mean_pct_rejected"].mean():.1f}%')
print(f'  Range: {summary_df["mean_pct_rejected"].min():.1f}% – {summary_df["mean_pct_rejected"].max():.1f}%')
```

```python
# =============================================================================
# 9b. Artifact Rejection Comparison: v2 vs v3
# =============================================================================
#
# This cell demonstrates the fix for artifact rejection.
# v2: Applied 150 µV threshold to raw data (with DC offset) → 95%+ rejection
# v3: Applied threshold to high-pass filtered data → reasonable rejection

TEST_SUBJECT = '10886'  # A no-earclip subject that had high rejection in v2

sub_info = SUBJECT_INFO[TEST_SUBJECT]
run_files = discover_runs(EEG_DIR, TEST_SUBJECT)

if 1 in run_files:
    files = run_files[1]
    nic2 = load_nic2_run(files['easy'], files['info'], files['imp'])
    
    # Get F4 data
    f4_raw = nic2['eeg'][:, EEG_CH_IDX[0]]
    
    # v2 approach: threshold on raw/reref data
    f4_reref = avg_rereference(nic2['eeg'])[:, 0]
    f4_uv_v2 = f4_reref / 1000
    v2_clean = np.abs(f4_uv_v2) < 150
    v2_pct_rejected = 100 * (1 - v2_clean.mean())
    
    # v3 approach: threshold on high-pass filtered data
    v3_clean, v3_pct_rejected, v3_info = reject_artifacts(f4_reref, threshold_uv=150)
    
    print(f'sub-{TEST_SUBJECT} Run 1 (baseline) — Artifact Rejection Comparison')
    print('=' * 60)
    print(f'v2 approach (threshold on raw): {v2_pct_rejected:.1f}% rejected')
    print(f'v3 approach (threshold on filtered): {v3_pct_rejected:.1f}% rejected')
    print(f'  - Amplitude rejection: {v3_info["pct_amplitude"]:.1f}%')
    print(f'  - Electrode pop rejection: {v3_info["pct_pops"]:.1f}%')
    print()
    print(f'Improvement: {v2_pct_rejected - v3_pct_rejected:.1f} percentage points more data retained')
else:
    print(f'Run 1 not found for sub-{TEST_SUBJECT}')
```

---
## Summary: v3 Changes

### 1. Artifact Rejection Fix
The v2 notebook rejected 92-99% of samples because the 150 µV threshold was applied to data with massive DC offsets (~350,000 µV for no-earclip participants). v3 applies the threshold to **high-pass filtered** data, which removes the DC offset first.

### 2. Reference-Aware Processing
Earclip and no-earclip participants now use different preprocessing parameters:
- **No-earclip**: Requires average re-referencing; higher stimulation detection threshold
- **Earclip**: No re-referencing needed; lower detection threshold (lower noise floor)

### 3. Adaptive Stimulation Detection
Instead of a fixed 120 dB threshold, detection now uses a **relative** threshold: active stimulation is detected when mid-run 6 Hz power exceeds the baseline estimate by X dB. This works for both earclip and no-earclip participants.

### 4. Electrode Pop Detection
Added detection for sudden amplitude jumps (like those seen in sub-10951), with 50ms exclusion windows around each pop.

### 5. Known Issues Documented
Protocol errors and data quality issues are now tracked in `SUBJECT_INFO['notes']` and propagated through the analysis.
