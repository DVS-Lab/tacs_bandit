# Feedback-Locked Theta Exploration

Exploratory notebook for detecting **feedback-locked theta (4–8 Hz)** in baseline (run‑1) Starstim EEG recordings.

## 1. Setup

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import signal
from tqdm import tqdm

plt.rcParams["figure.figsize"] = (8,4)
plt.rcParams["font.size"] = 12

FS = 500
THETA_BAND = (4,8)
FEEDBACK_DELAY = 2.5
EPOCH_WINDOW = (-2,3)

EEG_CHANNELS = ["F4","P4","P3"]
```

## 2. File Paths

```python
EEG_FILE = "20260312113121_sub-11318_Run 1.easy"
BEHAV_FILE = "sub-11318_ses-1_run-01_task-bandit_2026-03-12_11-29-26.csv"
```

## 3. Load EEG (.easy)

```python
print("Loading EEG...")

data = np.loadtxt(EEG_FILE)
print("Shape:", data.shape)

# Channels 5–7 are EEG
eeg = data[:,4:7]

timestamps = data[:,-1]

# Convert timestamps to seconds
time_sec = (timestamps - timestamps[0]) / 1000

duration = time_sec[-1]

print("Recording duration (s):", duration)
```

## 4. Quick EEG Diagnostics

### Raw EEG snippet

```python
start = 0
end = FS * 10

plt.figure()

for i,ch in enumerate(EEG_CHANNELS):
    plt.plot(time_sec[start:end],
             eeg[start:end,i] + i*200000,
             label=ch)

plt.title("Raw EEG (first 10s)")
plt.xlabel("Time (s)")
plt.legend()

plt.show()
```

### Power Spectrum

```python
plt.figure()

for i,ch in enumerate(EEG_CHANNELS):
    f, Pxx = signal.welch(eeg[:,i], FS, nperseg=4096)
    plt.semilogy(f, Pxx, label=ch)

plt.xlim(0,30)

plt.title("EEG Power Spectrum")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Power")

plt.legend()

plt.show()
```

## 5. Load Behavioral Data

```python
beh = pd.read_csv(BEHAV_FILE)

print("Trials:", len(beh))

# remove missed responses
beh = beh.dropna(subset=["rt"])

print("Trials after removing misses:", len(beh))
```

## 6. Compute Feedback Event Times

```python
beh["response_time"] = beh["trial_start_time"] + beh["rt"]
beh["feedback_time"] = beh["response_time"] + FEEDBACK_DELAY

feedback_times = beh["feedback_time"].values

print("Feedback events:", len(feedback_times))
```

## 7. Theta Power Helper Function

```python
def theta_power(x):

    b,a = signal.butter(4,
                        np.array(THETA_BAND)/(FS/2),
                        btype="band")

    x = signal.filtfilt(b,a,x)

    return np.mean(x**2)
```

## 8. Offset Search (EEG–Behavior Alignment)

```python
offsets = np.arange(-60,60,0.25)

scores = []

epoch_samples = int((EPOCH_WINDOW[1]-EPOCH_WINDOW[0])*FS)

for off in tqdm(offsets):

    trial_scores = []

    for t in feedback_times:

        eeg_time = t + off
        sample = int(eeg_time * FS)

        start = sample + int(EPOCH_WINDOW[0]*FS)
        end = sample + int(EPOCH_WINDOW[1]*FS)

        if start < 0 or end >= len(eeg):
            continue

        segment = eeg[start:end,0]

        mid = int(-EPOCH_WINDOW[0]*FS)

        pre = segment[mid-500:mid]
        post = segment[mid:mid+500]

        score = theta_power(post) - theta_power(pre)

        trial_scores.append(score)

    if len(trial_scores) > 5:
        scores.append(np.mean(trial_scores))
    else:
        scores.append(np.nan)
```

## 9. Plot Offset Search

```python
plt.figure()

plt.plot(offsets, scores)

plt.xlabel("Offset (s)")
plt.ylabel("Theta score")

plt.title("EEG–Behavior Alignment Search")

plt.show()

best_offset = offsets[np.nanargmax(scores)]

print("Best offset:", best_offset)
```

## 10. Extract Feedback-Locked Epochs

```python
epochs = []

for t in feedback_times:

    eeg_time = t + best_offset
    sample = int(eeg_time * FS)

    start = sample + int(EPOCH_WINDOW[0]*FS)
    end = sample + int(EPOCH_WINDOW[1]*FS)

    if start < 0 or end >= len(eeg):
        continue

    epochs.append(eeg[start:end,:])

epochs = np.array(epochs)

print("Epoch shape:", epochs.shape)
```

## 11. Spectrogram Around Feedback

```python
trial_avg = epochs.mean(axis=0)

f, t, S = signal.spectrogram(trial_avg[:,0],
                             FS,
                             nperseg=256,
                             noverlap=200)

plt.figure()

plt.pcolormesh(t + EPOCH_WINDOW[0],
               f,
               S,
               shading="gouraud")

plt.ylim(2,20)

plt.axvline(0,color="r")

plt.title("Feedback-locked spectrogram (F4)")

plt.xlabel("Time (s)")
plt.ylabel("Frequency (Hz)")

plt.show()
```

## 12. Theta Power Timecourse

```python
b,a = signal.butter(4,
                    np.array(THETA_BAND)/(FS/2),
                    btype="band")

theta_epochs = signal.filtfilt(b,a,epochs,axis=1)

theta_power_tc = np.mean(theta_epochs**2,axis=2)

time_axis = np.linspace(EPOCH_WINDOW[0],
                        EPOCH_WINDOW[1],
                        theta_power_tc.shape[1])

plt.figure()

plt.plot(time_axis,
         theta_power_tc.mean(axis=0))

plt.axvline(0,color="r")

plt.title("Theta power around feedback")

plt.xlabel("Time (s)")
plt.ylabel("Power")

plt.show()
```

## 13. Reward vs No‑Reward Theta

```python
reward = beh["reward"].values[:len(theta_power_tc)]

reward_theta = theta_power_tc[reward==1]
no_reward_theta = theta_power_tc[reward==0]

plt.figure()

plt.plot(time_axis,
         reward_theta.mean(axis=0),
         label="reward")

plt.plot(time_axis,
         no_reward_theta.mean(axis=0),
         label="no reward")

plt.axvline(0,color="k")

plt.legend()

plt.title("Theta: reward vs no reward")

plt.xlabel("Time (s)")
plt.ylabel("Power")

plt.show()
```
